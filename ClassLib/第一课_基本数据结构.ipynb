#几种回车方式
1. 使用 `\\` （在多行环境中）
$$
\begin{aligned}
a &= b + c \\
d &= e + f
\end{aligned}
$$
2. 使用 gather 环境（居中对齐）
$$
\begin{gather}
x = a + b \\
y = c + d
\end{gather}
$$
3. 使用 array 环境
$$
\begin{array}{l}
\text{第一行} \\
\text{第二行}
\end{array}
$$
4. 调整行间距 `\\`[间距]
$$
\begin{aligned}
a &= b \\[10pt]
c &= d
\end{aligned}
$$

注意：单独的 `$...$` 或 `$$...$$` 中不能直接使用 `\\`，必须配合 aligned、gather、array 等环境使用。

For example, you can calculator eigenvector with problem solving process:

For matrix: $A=\begin{bmatrix}	5 &6 &-3 \\	-1 &0 &1 \\	1 &2 &-1 \\\end{bmatrix}$

Let $B=\lambda\begin{bmatrix}1 &0 &0 \\0 &1 &0 \\0 &0 &1 \\\end{bmatrix}-\begin{bmatrix}	5 &6 &-3 \\	-1 &0 &1 \\	1 &2 &-1 \\\end{bmatrix}=\left[\begin{matrix}\lambda - 5 & -6 & 3\\1 & \lambda & -1\\-1 & -2 & \lambda + 1\end{matrix}\right]$

Its determinant is $latex(var["B"].doit().det()) = \lambda^{3} - 4 \lambda^{2} + 2 \lambda + 4$

Factor it: $latex(factor(var["B"].doit().det())) = \left(\lambda - 2\right) \left(\lambda^{2} - 2 \lambda - 2\right)$

Solve for eigenvalues $latex(solve(var["B"].doit().det())) = \left[ 2, \  1 - \sqrt{3}, \  1 + \sqrt{3}\right]$

Or via:$latex(var["A"].eigenvals()) = \left\{ 2 : 1, \  1 - \sqrt{3} : 1, \  1 + \sqrt{3} : 1\right\}$

Let $\lambda_1=2, \lambda_2=1-\sqrt{3}, \lambda_3=1+\sqrt{3}$

So $B_1=\lambda_1\begin{bmatrix}1 &0 &0 \\0 &1 &0 \\0 &0 &1 \\\end{bmatrix}-\begin{bmatrix}	5 &6 &-3 \\	-1 &0 &1 \\	1 &2 &-1 \\\end{bmatrix}=\left[\begin{matrix}-3 & -6 & 3\\1 & 2 & -1\\-1 & -2 & 3\end{matrix}\right]$

Assign it: $B_1=\left[\begin{matrix}-3 & -6 & 3\\1 & 2 & -1\\-1 & -2 & 3\end{matrix}\right]$

Simplify it by elementary row transformations:

$latex(var["B_1"].doit().rref()) = \left( \left[\begin{matrix}1 & 2 & 0\\0 & 0 & 1\\0 & 0 & 0\end{matrix}\right], \  \left( 0, \  2\right)\right)$

For the eigenvalue $\lambda_1=2$, its one eigenvector is $\xi_1=\begin{pmatrix}-2\\1\\0\end{pmatrix}$

So for the eigenvalue $\lambda_1=2$ its all eigenvectors are $k_1\xi_1 \ (k_1\neq 0, k_1\in P)$

In a same way, all eigenvalues and eigenvector are

$latex(var["A"].eigenvects()) = \left[ \left( 2, \  1, \  \left[ \left[\begin{matrix}-2\\1\\0\end{matrix}\right]\right]\right), \  \left( 1 - \sqrt{3}, \  1, \  \left[ \left[\begin{matrix}6 - 3 \sqrt{3}\\-2 + \sqrt{3}\\1\end{matrix}\right]\right]\right), \  \left( 1 + \sqrt{3}, \  1, \  \left[ \left[\begin{matrix}3 \sqrt{3} + 6\\-2 - \sqrt{3}\\1\end{matrix}\right]\right]\right)\right]$

`````python
#https://jupyter-notebook.readthedocs.io/en/latest/examples/Notebook/Importing%20Notebooks.html
import io, os,sys,types
from IPython import get_ipython
from nbformat import read
from IPython.core.interactiveshell import InteractiveShell
 
class NotebookFinder(object):
    """Module finder that locates Jupyter Notebooks"""
    def __init__(self):
        self.loaders = {}
 
    def find_module(self, fullname, path=None):
        nb_path = find_notebook(fullname, path)
        if not nb_path:
            return
 
        key = path
        if path:
            # lists aren't hashable
            key = os.path.sep.join(path)
 
        if key not in self.loaders:
            self.loaders[key] = NotebookLoader(path)
        return self.loaders[key]
 
def find_notebook(fullname, path=None):
    """find a notebook, given its fully qualified name and an optional path
    This turns "foo.bar" into "foo/bar.ipynb"
    and tries turning "Foo_Bar" into "Foo Bar" if Foo_Bar
    does not exist.
    """
    name = fullname.rsplit('.', 1)[-1]
    if not path:
        path = ['']
    for d in path:
        nb_path = os.path.join(d, name + ".ipynb")
        if os.path.isfile(nb_path):
            return nb_path
        # let import Notebook_Name find "Notebook Name.ipynb"
        nb_path = nb_path.replace("_", " ")
        if os.path.isfile(nb_path):
            return nb_path
 
class NotebookLoader(object):
    """Module Loader for Jupyter Notebooks"""
    def __init__(self, path=None):
        self.shell = InteractiveShell.instance()
        self.path = path
 
    def load_module(self, fullname):
        """import a notebook as a module"""
        path = find_notebook(fullname, self.path)
 
        print ("importing Jupyter notebook from %s" % path)
 
        # load the notebook object
        with io.open(path, 'r', encoding='utf-8') as f:
            nb = read(f, 4)
 
 
        # create the module and add it to sys.modules
        # if name in sys.modules:
        #    return sys.modules[name]
        mod = types.ModuleType(fullname)
        mod.__file__ = path
        mod.__loader__ = self
        mod.__dict__['get_ipython'] = get_ipython
        sys.modules[fullname] = mod
 
        # extra work to ensure that magics that would affect the user_ns
        # actually affect the notebook module's ns
        save_user_ns = self.shell.user_ns
        self.shell.user_ns = mod.__dict__
 
        try:
          for cell in nb.cells:
            if cell.cell_type == 'code':
                # transform the input to executable Python
                code = self.shell.input_transformer_manager.transform_cell(cell.source)
                # run the code in themodule
                exec(code, mod.__dict__)
        finally:
            self.shell.user_ns = save_user_ns
        return mod
sys.meta_path.append(NotebookFinder())
`````

### 定理证明
可以用其他定理证明
$\lim_{n \to \infty} |a_n - b_n| = |\lim_{n \to \infty} a_n - \lim_{n \to \infty} b(n)|$
吗？

证明:$\exists \{X_n\} \subset \mathring{U} (X_0),\{x_n\}->x_0 使得\lim_{n \to x_0} F(x_n) = F(x) \Rightarrow \lim_{x \to x_0} F(x_0) = F(x_0) $


求极限$\lim_{n \to \infty} \frac{ln(2n+1)}{n}$   
1) 首先，我们可以观察到这是一个 $\frac{\infty}{\infty}$ 型的不定式，因为当 $n \to \infty$ 时，分子和分母都趋向于无穷大。

2) 在这种情况下，我们可以应用洛必达法则。但是，直接应用洛必达法则会使问题变得复杂。因此，我们可以先进行一些代数变换。

3) 我们可以将分子 $\ln(2n+1)$ 改写为 $\ln(2n(1+\frac{1}{2n}))$：

   $\lim{n \to \infty} \frac{\ln(2n+1)}{n} = \lim{n \to \infty} \frac{\ln(2n(1+\frac{1}{2n}))}{n}$

4) 使用对数的性质，我们可以将其分解：

   $= \lim_{n \to \infty} \frac{\ln(2n) + \ln(1+\frac{1}{2n})}{n}$

5) 进一步分解：

   $= \lim_{n \to \infty} (\frac{\ln(2n)}{n} + \frac{\ln(1+\frac{1}{2n})}{n})$

6) 对于第一项 $\frac{\ln(2n)}{n}$：

   $\frac{\ln(2n)}{n} = \frac{\ln(2) + \ln(n)}{n} = \frac{\ln(2)}{n} + \frac{\ln(n)}{n}$

   当 $n \to \infty$ 时，$\frac{\ln(2)}{n} \to 0$，而 $\frac{\ln(n)}{n} \to 0$（这是一个著名的极限结果）。

7) 对于第二项 $\frac{\ln(1+\frac{1}{2n})}{n}$：

   当 $n \to \infty$ 时，$\frac{1}{2n} \to 0$，所以我们可以使用近似 $\ln(1+x) \approx x$ （当 $x$ 接近 0 时）。

   因此，$\frac{\ln(1+\frac{1}{2n})}{n} \approx \frac{\frac{1}{2n}}{n} = \frac{1}{2n^2} \to 0$ 当 $n \to \infty$。

8) 综合以上结果：

   $\lim{n \to \infty} \frac{\ln(2n+1)}{n} = \lim{n \to \infty} (\frac{\ln(2n)}{n} + \frac{\ln(1+\frac{1}{2n})}{n}) = 0 + 0 = 0$

结论
因此，我们证明了 $\lim_{n \to \infty} \frac{\ln(2n+1)}{n} = 0$

`````python
from manim import *

class LorenzAttractor(ThreeDScene):
    def construct(self):
        # 创建三维坐标轴
        axes = ThreeDAxes()
        self.add(axes)
        
        # 设置摄像机角度
        self.set_camera_orientation(phi=75 * DEGREES, theta=30 * DEGREES)
        
        # 时间步长和步数
        dt = 0.01
        num_steps = 10000
        
        # Lorenz 系统参数
        sigma = 10
        rho = 28
        beta = 8 / 3
        
        # 初始条件
        x = 0.01
        y = 0.0
        z = 0.0
        
        # 用于存储轨迹点
        points = []
        
        for _ in range(num_steps):
            # Lorenz 方程
            dx = sigma * (y - x) * dt
            dy = (x * (rho - z) - y) * dt
            dz = (x * y - beta * z) * dt
            
            x += dx
            y += dy
            z += dz
            
            points.append([x, y, z])
        
        # 将轨迹点转换为 Manim 的点
        trajectory = VMobject()
        trajectory.set_points_as_corners([axes.c2p(*point) for point in points])
        trajectory.set_color(YELLOW)
        
        # 动画显示轨迹
        self.play(Create(trajectory), run_time=8)
        self.wait()
`````
