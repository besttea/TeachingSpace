要素

```python
L=[]
a=[1,2,3,4]
 
#for i in range(len(a)):
 
for val_1 in a:   #   for(i=1;i<n;I++)
    for val_2 in a:
        for val_3 in a:
            if(val_1 == val_2 or val_1 == val_3 or val_2 == val_3):
                continue;
            else:
                temp=val_1*100+val_2*10+val_3
                if(temp%2 != 0):               
                    L.append(str(temp)) 
 
print(len(L)) 
print (L)
```

```python
from graphviz import Digraph

# 创建一个有向图对象
dot = Digraph(comment='Python Nested Loop Program')

# 添加节点
dot.node('A', '开始')
dot.node('B', '初始化L=[]; a=[1,2,3,4]')
dot.node('C', 'for val_1 in a')
dot.node('D', 'for val_2 in a')
dot.node('E', 'for val_3 in a')
dot.node('F', 'if val_1 == val_2 or val_1 == val_3 or val_2 == val_3')
dot.node('G', 'continue')
dot.node('H', 'L.append(str(val_1)+str(val_2)+str(val_3))')
dot.node('I', '打印len(L)和L')
dot.node('J', '结束')

# 添加边
dot.edge('A', 'B')
dot.edge('B', 'C')
dot.edge('C', 'D')
dot.edge('D', 'E')
dot.edge('E', 'F')
dot.edge('F', 'G', label='是', xlabel='True')
dot.edge('G', 'E')
dot.edge('F', 'H', label='否', xlabel='False')
dot.edge('H', 'E')
dot.edge('E', 'D')
dot.edge('D', 'C')
dot.edge('C', 'I')
dot.edge('I', 'J')

# 将图导出为文件
dot.render('flowchart', format='png', cleanup=True)
dot
```

```python
# 定义数字列表
numbers = [1, 2, 3, 4]

# 初始化计数器
count = 0

# 使用嵌套循环生成所有可能的三位数组合
for i in numbers:
    for j in numbers:
        for k in numbers:
            # 确保三个数字互不相同
            if i != j and j != k and i != k:
                # 打印满足条件的三位数
                print(f"{i}{j}{k}")
                # 增加计数器
                count += 1

# 打印总数
print(f"总共有 {count} 个互不相同且无重复数字的三位数。")
```

企业发放的奖金根据利润提成。
利润(I)低于或等于10万元时，奖金可提10%；
利润高于10万元，低于20万元时，低于10万元的部分
按10%提成，高于10万元的部分，可提成7.5%；
20万到40万之间时，高于20万元的部分，可提成5%；
40万到60万之间时高于40万元的部分，可提成3%；
60万到100万之间时，高于60万元的部分，可提成1.5%;
高于100万元时,超过100万元的部分按1%提成。
从键盘输入当月利润I，求应发放奖金总数？
 

```python

profit = 0
I = int(input("please input: "))
if(I<=10):
    profit = 0.1 * I
elif(I <= 20):
    profit = 10 *0.1 + (I - 10)*0.075
elif(I <=40):
    profit = 10 * 0.1 + (20 - 10)*0.075 + (I - 20)*0.05
elif(I <= 60):
    profit = 10 * 0.1 + (20 - 10)*0.075 + (40 - 20)*0.05 + (I - 40)*0.03
elif(I <= 100):
    profit = 10 * 0.1 + (20 - 10)*0.075 + (40 - 20)*0.05 + (60 - 40)*0.03 + (I - 60)*0.015
else : 
    profit = 10 * 0.1 + (20 - 10)*0.075 + (40 - 20)*0.05 + (60 - 40)*0.03 + (100 - 60)*0.015 + (I -100)*0.01
    
print ("profit=",profit)
        
```

```python
i = int(input('净利润:'))
arr = [1000000,600000,400000,200000,100000,0]
rat = [0.01,0.015,0.03,0.05,0.075,0.1]
r = 0
for idx in range(0,6):
    if i>arr[idx]:
        r+=(i-arr[idx])*rat[idx]#r=r+nnn
        print((i-arr[idx])*rat[idx])
        i=arr[idx]
print ("profit=",r)
```

```python
from graphviz import Digraph

# 创建一个有向图对象
dot = Digraph(comment='Profit Calculation Program')

# 添加节点
dot.node('A', '开始')
dot.node('B', '输入净利润 i')
dot.node('C', '初始化 arr 和 rat')
dot.node('D', 'r = 0')
dot.node('E', '遍历范围 0 至 5')
dot.node('F', '判断 i > arr[idx]',shape='parallelogram')
dot.node('G', '计算税额与累加\nr += (i - arr[idx]) * rat[idx]')
dot.node('H', '输出单项计算税额\nprint((i - arr[idx]) * rat[idx])')
dot.node('I', '更新净利润\ni = arr[idx]')
dot.node('J', '显示利润\nprint("profit=", r)')
dot.node('K', '结束')

# 添加边
dot.edge('A', 'B')
dot.edge('B', 'C')
dot.edge('C', 'D')
dot.edge('D', 'E')
dot.edge('E', 'F')
dot.edge('F', 'G', label='是')
dot.edge('F', 'E', label='否')
dot.edge('G', 'H')
dot.edge('H', 'I')
dot.edge('I', 'E')
dot.edge('E', 'J')
dot.edge('J', 'K')

# 将图导出为文件
dot.render('profit_flowchart', format='png', cleanup=True)
dot
```

```python
#Newton-Raphson for square root
# 
epsilon = 0.01
k = 24.0
guess = k/2.0
while abs(guess*guess -k) >= epsilon:
    guess = guess -(((guess**2) -k)/(2*guess))
    print('Square root of', k, 'is about', guess)
```

```python
def newton_raphson(f, df, x0, tol=1e-7, max_iter=1000):
    """
    牛顿-拉夫森法求解非线性方程 f(x) = 0

    参数:
    f: 函数对象，表示方程 f(x)
    df: 函数对象，表示 f(x) 的导数
    x0: 初始猜测值
    tol: 容差，迭代停止的阈值（默认 1e-7）
    max_iter: 最大迭代次数（默认 1000）

    返回:
    x: 方程的根（近似值）
    """
    x = x0
    for i in range(max_iter):
        fx = f(x)
        dfx = df(x)
        
        if dfx == 0:
           raise ValueError("导数为零，无法继续迭代")
       #   print("导数为零，无法继续迭代")
       #   break 
# 示例函数和它的导数
def f(x):
    return x**3 - x - 2
# 示例函数和它的导数
def f(x):
    return x**3 - x - 2

def df(x):
    return 3*x**2 - 1

# 初始猜测值
x0 = 1.0

# 使用牛顿-拉夫森法求解
try:
    root = newton_raphson(f, df, x0)
    print(f"方程的根为: {root}")
except ValueError as e:
    print(e)
```

```python
def generate_pascals_triangle(rows):
    triangle = []

    for i in range(rows):
        # 初始化当前行
        row = [1] * (i + 1)

        # 计算当前行中的中间值（不是第一个，也不是最后一个）
        for j in range(1, i):
            row[j] = triangle[i - 1][j - 1] + triangle[i - 1][j]

        triangle.append(row)

    return triangle

def print_pascals_triangle(triangle):
    max_width = len(' '.join(map(str, triangle[-1])))
    for row in triangle:
        row_str = ' '.join('*' if n else ' ' for n in row)
        print(row_str.center(max_width))

# 生成并打印10行杨辉三角
triangle = generate_pascals_triangle(10)
print_pascals_triangle(triangle)
```
